# Hyperparameter Tuning With Bayesian Optimization

>Link: https://scikit-optimize.github.io/stable/auto_examples/sklearn-gridsearchcv-replacement.html

This example demonstrates how to use bayesian optimization via skopt's bayessearchcv
to efficiently search over hyperparameter space. bayesian optimization models the
performance function with a probabilistic surrogate (often gaussian processes) and
selects the next hyperparameter values based on expected improvement.

>Important note: the `np.int` error fix has been applied by replacing `np.int` with `int` in the relevant file.
>Reference Link: https://scikit-optimize.github.io/stable/auto_examples/sklearn-gridsearchcv-replacement.html

```python
from skopt import BayesSearchCV
from sklearn.datasets import load_digits
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
```

```python
# load the digits dataset and split into training and test sets
X, y = load_digits(n_class=10, return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.75, test_size=.25, random_state=0)
```

```python
# define the bayesian search over hyperparameters for an svc
# 'log-uniform' indicates that the search is performed on the logarithmic scale,
# which is useful when parameters span several orders of magnitude.
opt = BayesSearchCV(
    SVC(),
    {
        'C': (1e-6, 1e+6, 'log-uniform'),
        'gamma': (1e-6, 1e+1, 'log-uniform'),
        'degree': (1, 8),  # integer valued parameter
        'kernel': ['linear', 'poly', 'rbf'],  # categorical parameter
    },
    n_iter=32,
    cv=3
)
```

```python
# fit the bayesian search model on the training data
opt.fit(X_train, y_train)

# print the best hyperparameter values, best cross-validation score, and test score
print("best params: %s" % str(opt.best_params_))
print("val. score: %s" % opt.best_score_)
print("test score: %s" % opt.score(X_test, y_test))
```

## Counting total iterations that will be used to explore all subspaces

In this example, the search_spaces parameter can include subspaces with a specific number of iterations.
The total iterations is the sum over all specified subspaces.

```python
from skopt import BayesSearchCV
from sklearn.datasets import load_iris
from sklearn.svm import SVC

# load the iris dataset
X, y = load_iris(return_X_y=True)

# define a bayesian search with two subspaces; one subspace has a specified number of iterations
opt = BayesSearchCV(
    SVC(),
    search_spaces=[
        ({'C': (0.1, 1.0)}, 19),  # 19 iterations for this subspace
        {'gamma':(0.1, 1.0)}
    ],
    n_iter=23
)

print('total iterations=', opt.total_iterations)
```
